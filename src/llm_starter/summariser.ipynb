```python
import os 
from dotenv import load_dotenv
from scraper import fetch_website_contents
from IPython.display import Markdown, display
from openai import OpenAI 

load_dotenv(override=True)
api_key = os.getenv('GOOGLE_API_KEY')
gemini_base_url = os.getenv('GEMINI_BASE_URL')
gemini_model = os.getenv('GOOGLE_MODEL')

if not api_key:
    print("No API key was found")


# Define our system prompt "
system_prompt = """
You are a snarky assistant that analyzes the contents of a website,
and provides a short, snarky, humorous summary, ignoring text that might be navigation related.
Respond in markdown. Do not wrap the markdown in a code block - respond just with the markdown.
"""

# Define our user prompt
user_prompt_prefix = """
Here are the contents of a website.
Provide a short summary of this website.
If it includes news or announcements, then summarize these too.

"""

gemini_client = OpenAI(base_url=gemini_base_url, api_key=api_key)

def construct_messages(website):
    contents = fetch_website_contents(website)
    messages = [
        {"role": "system", "content": system_prompt},
        {"role": "user", "content": user_prompt_prefix + contents}
    ]
    return messages

def summarise(url):
    response = gemini_client.chat.completions.create(
        model=gemini_model, 
        messages=construct_messages(url)
    )
    return response.choices[0].message.content

def display_summary(url):
    summary = summarise(url)
    display(Markdown(summary))

summarise("https://www.canva.in/")

display_summary("https://www.canva.in/")
```
